# **Ejercicio 1 – Calidad del software.**
La clase Producto presenta características de calidad como legibilidad, mantenibilidad y reutilización. La legibilidad se facilita utilizando nombres claros para los atributos y métodos. La mantenibilidad se logra separando las diferentes operaciones en métodos, como mostrar_info(), actualizar_precio() y vender(). También existe reutilización, porque la misma clase permite crear diferentes productos sin repetir toda la estructura. Una debilidad de la versión inicial era que los datos podían modificarse directamente sin suficiente control. Para corregir esta situación se aplicó encapsulamiento utilizando atributos privados y propiedades, permitiendo validar valores como el precio y la cantidad antes de modificarlos.

```python
#Primera parte  del codigo
from abc import ABC, abstractmethod

# CLASE ABSTRACTA PRODUCTO
class Producto(ABC):

    total_productos = 0

    def __init__(self, nombre, precio, cantidad, categoria):
        # ENCAPSULAMIENTO:
        # Los atributos importantes se protegen
        # utilizando atributos privados.
        self.__nombre = nombre
        self.__precio = precio
        self.__cantidad = cantidad
        self.categoria = categoria

        Producto.total_productos += 1

    @property
    def nombre(self):
        return self.__nombre

    @property
    def precio(self):
        return self.__precio

    @precio.setter
    def precio(self, nuevo_precio):
        if nuevo_precio <= 0:
            raise ValueError("El precio debe ser mayor que cero.")
        self.__precio = nuevo_precio

    @property
    def cantidad(self):
        return self.__cantidad

    @cantidad.setter
    def cantidad(self, nueva_cantidad):
        if nueva_cantidad < 0:
            raise ValueError("La cantidad no puede ser negativa.")
        self.__cantidad = nueva_cantidad

    # ABSTRACCIÓN:
    # La clase Producto representa las características
    # generales de cualquier producto de la tiendita.
    @abstractmethod
    def tipo_producto(self):
        pass

    def vender(self, cantidad=1):
        if cantidad <= 0:
            raise ValueError("La cantidad debe ser mayor que cero.")
        if cantidad > self.cantidad:
            print("No hay suficiente producto disponible.")
            return
        self.cantidad -= cantidad
        print(f"Venta realizada: {cantidad} unidad(es).")

    # SOBRECARGA SIMULADA:
    def actualizar_precio(self, nuevo_precio, descuento=0):
        precio_final = nuevo_precio - descuento
        if precio_final <= 0:
            raise ValueError("El precio debe ser mayor que cero.")
        self.precio = precio_final

    def mostrar_info(self):
        return (
            f"Producto: {self.nombre}\n"
            f"Precio: ${self.precio:.2f}\n"
            f"Cantidad: {self.cantidad}\n"
            f"Categoría: {self.categoria}"
        )
```

```python
# Agregamos los metodos con indentacion corregida
def vender(self, cantidad=1):
    if cantidad <= 0:
        raise ValueError("La cantidad debe ser mayor que cero.")

    if cantidad > self.cantidad:
        print("No hay suficiente producto disponible.")
        return

    self.cantidad -= cantidad
    print(f"Venta realizada: {cantidad} unidad(es).")

# SOBRECARGA SIMULADA:
# Python no permite sobrecarga real de métodos.
# Se simula utilizando un argumento por defecto.
def actualizar_precio(self, nuevo_precio, descuento=0):
    precio_final = nuevo_precio - descuento

    if precio_final <= 0:
        raise ValueError("El precio debe ser mayor que cero.")

    self.precio = precio_final

def mostrar_info(self):
    return (
        f"Producto: {self.nombre}\n"
        f"Precio: ${self.precio:.2f}\n"
        f"Cantidad: {self.cantidad}\n"
        f"Categoría: {self.categoria}"
    )
```

```python
#Agregamos Herencia
class ProductoAlimento(Producto):

    def __init__(self, nombre, precio, cantidad, categoria, marca):
        super().__init__(nombre, precio, cantidad, categoria)
        self.marca = marca

    # HERENCIA:
    # ProductoAlimento hereda de Producto.

    # POLIMORFISMO:
    # Se sobrescribe el método mostrar_info().
    def mostrar_info(self):

        return (
            f"ALIMENTO\n"
            f"Producto: {self.nombre}\n"
            f"Marca: {self.marca}\n"
            f"Precio: ${self.precio:.2f}\n"
            f"Cantidad: {self.cantidad}\n"
            f"Categoría: {self.categoria}"
        )

    def tipo_producto(self):
        return "Alimento"
```

```python
class ProductoBebida(Producto):

    def __init__(self, nombre, precio, cantidad, categoria, tamaño):
        super().__init__(nombre, precio, cantidad, categoria)
        self.tamaño = tamaño

    # POLIMORFISMO:
    # También sobrescribe mostrar_info().
    def mostrar_info(self):

        return (
            f"BEBIDA\n"
            f"Producto: {self.nombre}\n"
            f"Tamaño: {self.tamaño}\n"
            f"Precio: ${self.precio:.2f}\n"
            f"Cantidad: {self.cantidad}\n"
            f"Categoría: {self.categoria}"
        )

    def tipo_producto(self):
        return "Bebida"
```

```python
#Creamos los Objetos
producto1 = ProductoAlimento(
    "Galletas Granados",
    1.25,
    20,
    "Alimentos",
    "Marca A"
)

producto2 = ProductoBebida(
    "Coca Cola",
    1.25,
    15,
    "Bebidas",
    "600 ml"
)
```

```python
#Demostracion de polimorfismo
productos = [producto1, producto2]

print("===== POLIMORFISMO =====")

for producto in productos:
    print(producto.mostrar_info())
    print("----------------------")
```
